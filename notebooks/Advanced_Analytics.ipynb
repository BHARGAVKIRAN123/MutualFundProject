```python
# ============================================================
# BLUESTOCK MUTUAL FUND CAPSTONE
# ADVANCED ANALYTICS + RISK METRICS
# ============================================================

# ============================================================
# 1. IMPORTS AND PROJECT PATHS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
from IPython.display import display, Markdown

# Project root
PROJECT_ROOT = Path.cwd().parent

DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
DASHBOARD = PROJECT_ROOT / "dashboard"
REPORTS = PROJECT_ROOT / "reports"
SCRIPTS = PROJECT_ROOT / "scripts"

# Create folders if they do not exist
DASHBOARD.mkdir(parents=True, exist_ok=True)
REPORTS.mkdir(parents=True, exist_ok=True)
SCRIPTS.mkdir(parents=True, exist_ok=True)

print("Advanced Analytics environment ready.")
print("Project root:", PROJECT_ROOT)


# ============================================================
# 2. HELPER FUNCTION TO FIND DATASETS
# ============================================================

def find_csv(required_columns, preferred_names=None):
    """
    Find a CSV containing all required columns.
    Searches processed first, then raw.
    """

    required_columns = set(required_columns)

    candidates = []

    for folder in [DATA_PROCESSED, DATA_RAW]:

        if not folder.exists():
            continue

        for file in folder.glob("*.csv"):

            candidates.append(file)

    # Prefer exact filenames when supplied
    if preferred_names:

        for preferred in preferred_names:

            for file in candidates:

                if file.name.lower() == preferred.lower():

                    try:
                        cols = set(
                            pd.read_csv(
                                file,
                                nrows=2
                            ).columns
                        )

                        if required_columns.issubset(cols):
                            return file

                    except Exception:
                        pass

    # Otherwise search by columns
    for file in candidates:

        try:

            cols = set(
                pd.read_csv(
                    file,
                    nrows=2
                ).columns
            )

            if required_columns.issubset(cols):
                return file

        except Exception:
            continue

    return None


# ============================================================
# 3. LOAD NAV DATA
# ============================================================

nav_file = find_csv(
    [
        "amfi_code",
        "date",
        "nav"
    ],
    preferred_names=[
        "02_nav_history_clean.csv",
        "02_nav_history.csv"
    ]
)

if nav_file is None:
    raise FileNotFoundError(
        "NAV dataset not found."
    )

nav = pd.read_csv(nav_file)

print("NAV file:")
print(nav_file)

print("\nNAV shape:", nav.shape)

print("\nNAV columns:")
print(nav.columns.tolist())

display(nav.head())


# ============================================================
# 4. LOAD PERFORMANCE DATA
# ============================================================

performance_file = find_csv(
    [
        "amfi_code",
        "scheme_name",
        "fund_house",
        "sharpe_ratio",
        "risk_grade",
        "aum_crore"
    ],
    preferred_names=[
        "07_scheme_performance_clean.csv",
        "07_scheme_performance.csv"
    ]
)

if performance_file is None:
    raise FileNotFoundError(
        "Scheme performance dataset not found."
    )

performance = pd.read_csv(
    performance_file
)

print("\nPerformance file:")
print(performance_file)

print("\nPerformance shape:")
print(performance.shape)

print("\nPerformance columns:")
print(performance.columns.tolist())

display(performance.head())


# ============================================================
# 5. LOAD INVESTOR TRANSACTIONS
# ============================================================

transactions_file = find_csv(
    [
        "investor_id",
        "transaction_date",
        "amfi_code",
        "transaction_type",
        "amount_inr"
    ],
    preferred_names=[
        "08_investor_transactions_clean.csv",
        "08_investor_transactions.csv"
    ]
)

if transactions_file is None:
    raise FileNotFoundError(
        "Investor transactions dataset not found."
    )

transactions = pd.read_csv(
    transactions_file
)

print("\nTransactions file:")
print(transactions_file)

print("\nTransactions shape:")
print(transactions.shape)

print("\nTransaction columns:")
print(transactions.columns.tolist())

display(transactions.head())


# ============================================================
# 6. LOAD HOLDINGS DATA
# ============================================================

holdings_file = find_csv(
    [
        "amfi_code",
        "stock_symbol",
        "sector",
        "weight_pct"
    ]
)

if holdings_file is None:
    raise FileNotFoundError(
        "Portfolio holdings dataset not found."
    )

holdings = pd.read_csv(
    holdings_file
)

print("\nHoldings file:")
print(holdings_file)

print("\nHoldings shape:")
print(holdings.shape)

print("\nHoldings columns:")
print(holdings.columns.tolist())

display(holdings.head())


# ============================================================
# 7. PREPARE NAV DATA
# ============================================================

nav["date"] = pd.to_datetime(
    nav["date"],
    errors="coerce"
)

nav["nav"] = pd.to_numeric(
    nav["nav"],
    errors="coerce"
)

nav = nav.dropna(
    subset=[
        "date",
        "amfi_code",
        "nav"
    ]
).copy()

nav = nav.sort_values(
    [
        "amfi_code",
        "date"
    ]
)

# Daily return
nav["daily_return"] = (
    nav
    .groupby("amfi_code")["nav"]
    .pct_change()
)

print(
    "\nNumber of schemes:",
    nav["amfi_code"].nunique()
)

print(
    "Number of return observations:",
    nav["daily_return"].notna().sum()
)


# ============================================================
# 8. HISTORICAL VaR + CVaR
# ============================================================

# 95% Historical VaR:
# 5th percentile of daily returns
#
# CVaR:
# Average return of observations below/equal to VaR

var_results = []

for amfi_code, group in nav.groupby("amfi_code"):

    returns = (
        group["daily_return"]
        .dropna()
    )

    if len(returns) == 0:
        continue

    var_95 = returns.quantile(0.05)

    tail_returns = returns[
        returns <= var_95
    ]

    cvar_95 = (
        tail_returns.mean()
        if len(tail_returns) > 0
        else np.nan
    )

    var_results.append(
        {
            "amfi_code": amfi_code,
            "observations": len(returns),
            "VaR_95_pct": var_95 * 100,
            "CVaR_95_pct": cvar_95 * 100
        }
    )

var_cvar = pd.DataFrame(
    var_results
)

# Add scheme information
performance_lookup = performance[
    [
        "amfi_code",
        "scheme_name",
        "fund_house",
        "category",
        "risk_grade"
    ]
].drop_duplicates(
    subset=["amfi_code"]
)

var_cvar = var_cvar.merge(
    performance_lookup,
    on="amfi_code",
    how="left"
)

var_cvar = var_cvar.sort_values(
    "VaR_95_pct"
)

print("\nHistorical VaR / CVaR:")
display(
    var_cvar.head(20)
)


# ============================================================
# 9. SAVE VaR / CVaR REPORT
# ============================================================

var_report_file = (
    REPORTS / "var_cvar_report.csv"
)

var_cvar.to_csv(
    var_report_file,
    index=False
)

print(
    "\nVaR/CVaR report saved:"
)

print(var_report_file)


# ============================================================
# 10. TOP 5 FUNDS FOR ROLLING SHARPE
# ============================================================

# Prefer highest-AUM funds from performance.
# If that fails, use funds with the most NAV observations.

if "aum_crore" in performance.columns:

    performance["aum_crore"] = pd.to_numeric(
        performance["aum_crore"],
        errors="coerce"
    )

    top_5_codes = (
        performance
        .dropna(subset=["aum_crore"])
        .sort_values(
            "aum_crore",
            ascending=False
        )
        ["amfi_code"]
        .drop_duplicates()
        .head(5)
        .tolist()
    )

else:

    top_5_codes = (
        nav.groupby("amfi_code")
        .size()
        .sort_values(
            ascending=False
        )
        .head(5)
        .index
        .tolist()
    )

print(
    "\nSelected 5 funds:"
)

print(top_5_codes)


# ============================================================
# 11. ROLLING 90-DAY SHARPE
# ============================================================

rolling_sharpe_frames = []

for code in top_5_codes:

    fund = nav[
        nav["amfi_code"] == code
    ].copy()

    fund = fund.sort_values("date")

    # Manager's formula:
    # rolling mean / rolling std * sqrt(252)

    fund["rolling_sharpe_90"] = (
        fund["daily_return"]
        .rolling(90)
        .mean()
        /
        fund["daily_return"]
        .rolling(90)
        .std()
        *
        np.sqrt(252)
    )

    rolling_sharpe_frames.append(
        fund[
            [
                "date",
                "amfi_code",
                "rolling_sharpe_90"
            ]
        ]
    )

rolling_sharpe = pd.concat(
    rolling_sharpe_frames,
    ignore_index=True
)

# Add scheme names
rolling_sharpe = rolling_sharpe.merge(
    performance[
        [
            "amfi_code",
            "scheme_name"
        ]
    ].drop_duplicates(
        subset=["amfi_code"]
    ),
    on="amfi_code",
    how="left"
)

display(
    rolling_sharpe.head()
)


# ============================================================
# 12. ROLLING SHARPE CHART
# ============================================================

plt.figure(figsize=(15, 8))

for code in top_5_codes:

    fund_data = rolling_sharpe[
        rolling_sharpe["amfi_code"] == code
    ]

    if fund_data.empty:
        continue

    name = fund_data["scheme_name"].iloc[0]

    plt.plot(
        fund_data["date"],
        fund_data["rolling_sharpe_90"],
        label=str(name)
    )

plt.axhline(
    0,
    linestyle="--",
    linewidth=1
)

plt.title(
    "90-Day Rolling Sharpe Ratio - Top 5 Funds",
    fontsize=16
)

plt.xlabel("Date")
plt.ylabel("Rolling Sharpe Ratio")

plt.legend(
    fontsize=8,
    loc="best"
)

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()

rolling_sharpe_chart = (
    DASHBOARD / "rolling_sharpe_chart.png"
)

plt.savefig(
    rolling_sharpe_chart,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(
    "\nRolling Sharpe chart saved:"
)

print(rolling_sharpe_chart)


# ============================================================
# 13. PREPARE TRANSACTION DATA
# ============================================================

transactions["transaction_date"] = pd.to_datetime(
    transactions["transaction_date"],
    errors="coerce"
)

transactions["amount_inr"] = pd.to_numeric(
    transactions["amount_inr"],
    errors="coerce"
)

transactions = transactions.dropna(
    subset=[
        "investor_id",
        "transaction_date",
        "amount_inr",
        "amfi_code"
    ]
).copy()

transactions["transaction_type_clean"] = (
    transactions["transaction_type"]
    .astype(str)
    .str.strip()
    .str.upper()
)

print(
    "\nTransaction types:"
)

print(
    transactions["transaction_type_clean"]
    .value_counts()
)


# ============================================================
# 14. IDENTIFY SIP TRANSACTIONS
# ============================================================

sip_transactions = transactions[
    transactions["transaction_type_clean"]
    .str.contains(
        "SIP",
        na=False
    )
].copy()

print(
    "\nSIP transaction records:",
    len(sip_transactions)
)


# ============================================================
# 15. INVESTOR COHORT ANALYSIS
# ============================================================

# First transaction year
first_transaction = (
    transactions
    .groupby("investor_id")["transaction_date"]
    .min()
    .reset_index()
)

first_transaction["cohort_year"] = (
    first_transaction["transaction_date"]
    .dt.year
)

# Attach cohort year
cohort_transactions = transactions.merge(
    first_transaction[
        [
            "investor_id",
            "cohort_year"
        ]
    ],
    on="investor_id",
    how="left"
)

# SIP transactions with cohort
cohort_sip = sip_transactions.merge(
    first_transaction[
        [
            "investor_id",
            "cohort_year"
        ]
    ],
    on="investor_id",
    how="left"
)

# Average SIP amount by cohort
cohort_summary = (
    cohort_sip
    .groupby("cohort_year")
    .agg(
        avg_sip_amount_inr=(
            "amount_inr",
            "mean"
        ),
        total_sip_invested_inr=(
            "amount_inr",
            "sum"
        ),
        unique_investors=(
            "investor_id",
            "nunique"
        )
    )
    .reset_index()
)

# Total invested from all transaction types
total_invested = (
    cohort_transactions
    .groupby("cohort_year")
    .agg(
        total_invested_all_transactions_inr=(
            "amount_inr",
            "sum"
        )
    )
    .reset_index()
)

cohort_summary = cohort_summary.merge(
    total_invested,
    on="cohort_year",
    how="left"
)

# Top fund preference by SIP count
fund_preference = (
    cohort_sip
    .groupby(
        [
            "cohort_year",
            "amfi_code"
        ]
    )
    .agg(
        sip_transactions=(
            "investor_id",
            "size"
        ),
        investors=(
            "investor_id",
            "nunique"
        ),
        sip_amount_inr=(
            "amount_inr",
            "sum"
        )
    )
    .reset_index()
)

# Top fund within each cohort
fund_preference["rank"] = (
    fund_preference
    .groupby("cohort_year")[
        "sip_transactions"
    ]
    .rank(
        ascending=False,
        method="first"
    )
)

top_cohort_funds = fund_preference[
    fund_preference["rank"] == 1
].copy()

top_cohort_funds = top_cohort_funds.merge(
    performance[
        [
            "amfi_code",
            "scheme_name"
        ]
    ].drop_duplicates(
        subset=["amfi_code"]
    ),
    on="amfi_code",
    how="left"
)

top_cohort_funds = top_cohort_funds[
    [
        "cohort_year",
        "amfi_code",
        "scheme_name",
        "sip_transactions",
        "investors",
        "sip_amount_inr"
    ]
]

cohort_summary = cohort_summary.merge(
    top_cohort_funds[
        [
            "cohort_year",
            "scheme_name"
        ]
    ],
    on="cohort_year",
    how="left"
)

cohort_summary = cohort_summary.rename(
    columns={
        "scheme_name": "top_fund_preference"
    }
)

print(
    "\nInvestor cohort analysis:"
)

display(cohort_summary)


# ============================================================
# 16. SAVE COHORT REPORT
# ============================================================

cohort_report_file = (
    REPORTS / "investor_cohort_analysis.csv"
)

cohort_summary.to_csv(
    cohort_report_file,
    index=False
)

print(
    "Cohort report saved:"
)

print(cohort_report_file)


# ============================================================
# 17. SIP CONTINUITY ANALYSIS
# ============================================================

# Need investors with at least 6 SIP transactions

sip_dates = (
    sip_transactions
    [
        [
            "investor_id",
            "transaction_date"
        ]
    ]
    .drop_duplicates()
    .sort_values(
        [
            "investor_id",
            "transaction_date"
        ]
    )
)

sip_counts = (
    sip_dates
    .groupby("investor_id")
    .size()
)

eligible_investors = sip_counts[
    sip_counts >= 6
].index

eligible_sip_dates = sip_dates[
    sip_dates["investor_id"]
    .isin(eligible_investors)
].copy()

# Gap between consecutive SIP dates
eligible_sip_dates["gap_days"] = (
    eligible_sip_dates
    .groupby("investor_id")[
        "transaction_date"
    ]
    .diff()
    .dt.days
)

# Average gap per investor
continuity = (
    eligible_sip_dates
    .groupby("investor_id")
    .agg(
        sip_transactions=(
            "transaction_date",
            "size"
        ),
        avg_gap_days=(
            "gap_days",
            "mean"
        ),
        max_gap_days=(
            "gap_days",
            "max"
        )
    )
    .reset_index()
)

# At-risk if any gap > 35 days
continuity["at_risk"] = (
    continuity["max_gap_days"] > 35
)

print(
    "\nSIP continuity analysis:"
)

display(
    continuity.head(20)
)

print(
    "\nEligible investors:",
    len(continuity)
)

print(
    "At-risk investors:",
    continuity["at_risk"].sum()
)


# ============================================================
# 18. SIP CONTINUITY SUMMARY
# ============================================================

continuity_rate = (
    (~continuity["at_risk"]).mean() * 100
    if len(continuity) > 0
    else np.nan
)

print(
    f"\nSIP continuity rate: "
    f"{continuity_rate:.2f}%"
)

print(
    f"At-risk rate: "
    f"{100 - continuity_rate:.2f}%"
)


# ============================================================
# 19. SAVE SIP CONTINUITY REPORT
# ============================================================

continuity_report_file = (
    REPORTS / "sip_continuity_analysis.csv"
)

continuity.to_csv(
    continuity_report_file,
    index=False
)

print(
    "Continuity report saved:"
)

print(continuity_report_file)


# ============================================================
# 20. SIMPLE FUND RECOMMENDER
# ============================================================

# Clean risk grade
performance["risk_grade_clean"] = (
    performance["risk_grade"]
    .astype(str)
    .str.strip()
    .str.lower()
)

performance["sharpe_ratio"] = pd.to_numeric(
    performance["sharpe_ratio"],
    errors="coerce"
)

recommendation_rows = []

requested_risks = [
    "Low",
    "Moderate",
    "High"
]

for requested_risk in requested_risks:

    risk_key = requested_risk.lower()

    matching = performance[
        performance["risk_grade_clean"]
        == risk_key
    ].copy()

    # Fallback:
    # if exact matching risk grade is absent,
    # use text matching carefully.
    if matching.empty:

        if requested_risk == "Low":

            matching = performance[
                performance[
                    "risk_grade_clean"
                ].str.contains(
                    "low",
                    na=False
                )
                &
                ~performance[
                    "risk_grade_clean"
                ].str.contains(
                    "moderate",
                    na=False
                )
            ].copy()

        elif requested_risk == "Moderate":

            matching = performance[
                performance[
                    "risk_grade_clean"
                ].str.contains(
                    "moderate",
                    na=False
                )
            ].copy()

        elif requested_risk == "High":

            matching = performance[
                performance[
                    "risk_grade_clean"
                ].str.contains(
                    "high",
                    na=False
                )
            ].copy()

    matching = matching.dropna(
        subset=["sharpe_ratio"]
    )

    matching = matching.sort_values(
        "sharpe_ratio",
        ascending=False
    ).head(3)

    for rank, (_, row) in enumerate(
        matching.iterrows(),
        start=1
    ):

        recommendation_rows.append(
            {
                "requested_risk": requested_risk,
                "rank": rank,
                "amfi_code": row["amfi_code"],
                "scheme_name": row["scheme_name"],
                "fund_house": row["fund_house"],
                "risk_grade": row["risk_grade"],
                "sharpe_ratio": row["sharpe_ratio"]
            }
        )

recommendations = pd.DataFrame(
    recommendation_rows
)

print(
    "\nFund recommendations:"
)

display(recommendations)


# ============================================================
# 21. SAVE RECOMMENDER TABLE
# ============================================================

recommendation_file = (
    REPORTS / "fund_recommendations.csv"
)

recommendations.to_csv(
    recommendation_file,
    index=False
)

print(
    "Recommendation report saved:"
)

print(recommendation_file)


# ============================================================
# 22. CREATE recommender.py
# ============================================================

recommender_code = r'''
import pandas as pd
from pathlib import Path


def recommend_funds(
    performance_file,
    risk_appetite,
    top_n=3
):
    """
    Recommend top funds by Sharpe ratio
    within the requested risk grade.

    Parameters
    ----------
    performance_file : str or Path
        Path to scheme performance CSV.

    risk_appetite : str
        Low / Moderate / High

    top_n : int
        Number of funds to return.

    Returns
    -------
    pandas.DataFrame
    """

    df = pd.read_csv(
        performance_file
    )

    df["risk_grade_clean"] = (
        df["risk_grade"]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    df["sharpe_ratio"] = pd.to_numeric(
        df["sharpe_ratio"],
        errors="coerce"
    )

    risk = str(
        risk_appetite
    ).strip().lower()

    # Exact match first
    result = df[
        df["risk_grade_clean"] == risk
    ].copy()

    # Fallback matching
    if result.empty:

        if risk == "low":

            result = df[
                df["risk_grade_clean"]
                .str.contains(
                    "low",
                    na=False
                )
                &
                ~df[
                    "risk_grade_clean"
                ].str.contains(
                    "moderate",
                    na=False
                )
            ].copy()

        elif risk == "moderate":

            result = df[
                df["risk_grade_clean"]
                .str.contains(
                    "moderate",
                    na=False
                )
            ].copy()

        elif risk == "high":

            result = df[
                df["risk_grade_clean"]
                .str.contains(
                    "high",
                    na=False
                )
            ].copy()

    result = result.dropna(
        subset=["sharpe_ratio"]
    )

    result = (
        result
        .sort_values(
            "sharpe_ratio",
            ascending=False
        )
        .head(top_n)
    )

    return result[
        [
            "amfi_code",
            "scheme_name",
            "fund_house",
            "risk_grade",
            "sharpe_ratio"
        ]
    ]


if __name__ == "__main__":

    project_root = Path(__file__).resolve().parents[1]

    performance_file = (
        project_root
        / "data"
        / "processed"
        / "07_scheme_performance_clean.csv"
    )

    print("\nLow Risk:")
    print(
        recommend_funds(
            performance_file,
            "Low"
        ).to_string(index=False)
    )

    print("\nModerate Risk:")
    print(
        recommend_funds(
            performance_file,
            "Moderate"
        ).to_string(index=False)
    )

    print("\nHigh Risk:")
    print(
        recommend_funds(
            performance_file,
            "High"
        ).to_string(index=False)
    )
'''

recommender_file = (
    SCRIPTS / "recommender.py"
)

with open(
    recommender_file,
    "w",
    encoding="utf-8"
) as f:

    f.write(recommender_code)

print(
    "\nrecommender.py created:"
)

print(recommender_file)


# ============================================================
# 23. SECTOR HHI CONCENTRATION
# ============================================================

holdings["weight_pct"] = pd.to_numeric(
    holdings["weight_pct"],
    errors="coerce"
)

holdings["portfolio_date"] = pd.to_datetime(
    holdings["portfolio_date"],
    errors="coerce"
)

hhi_data = holdings.dropna(
    subset=[
        "amfi_code",
        "sector",
        "weight_pct"
    ]
).copy()

# Sector weight per fund
fund_sector_weights = (
    hhi_data
    .groupby(
        [
            "amfi_code",
            "sector"
        ]
    )["weight_pct"]
    .sum()
    .reset_index()
)

# HHI formula:
# Sum of squared sector weights.
#
# We use decimal weights:
# 40% -> 0.40
# HHI = sum(weight_decimal^2)

fund_sector_weights["weight_decimal"] = (
    fund_sector_weights["weight_pct"] / 100
)

hhi_results = (
    fund_sector_weights
    .groupby("amfi_code")
    .apply(
        lambda x: (
            x["weight_decimal"] ** 2
        ).sum(),
        include_groups=False
    )
    .reset_index(
        name="sector_hhi"
    )
)

# Convert to 0-10000 HHI scale
hhi_results["sector_hhi_10000"] = (
    hhi_results["sector_hhi"] * 10000
)

# Add fund details
hhi_results = hhi_results.merge(
    performance[
        [
            "amfi_code",
            "scheme_name",
            "fund_house",
            "category"
        ]
    ].drop_duplicates(
        subset=["amfi_code"]
    ),
    on="amfi_code",
    how="left"
)

# Identify equity funds
hhi_results["category_clean"] = (
    hhi_results["category"]
    .astype(str)
    .str.lower()
)

equity_hhi = hhi_results[
    hhi_results["category_clean"]
    .str.contains(
        "equity",
        na=False
    )
].copy()

equity_hhi = equity_hhi.sort_values(
    "sector_hhi_10000",
    ascending=False
)

print(
    "\nSector HHI for equity funds:"
)

display(
    equity_hhi[
        [
            "amfi_code",
            "scheme_name",
            "fund_house",
            "category",
            "sector_hhi_10000"
        ]
    ].head(20)
)


# ============================================================
# 24. HHI INTERPRETATION
# ============================================================

print("\nHHI interpretation:")

print(
    "Lower HHI = more diversified across sectors."
)

print(
    "Higher HHI = more concentrated sector exposure."
)


# ============================================================
# 25. SAVE HHI REPORT
# ============================================================

hhi_report_file = (
    REPORTS / "sector_hhi_report.csv"
)

hhi_results.to_csv(
    hhi_report_file,
    index=False
)

print(
    "\nHHI report saved:"
)

print(hhi_report_file)


# ============================================================
# 26. TOP HHI FUNDS
# ============================================================

if not equity_hhi.empty:

    print(
        "\nTop 10 most concentrated equity funds:"
    )

    display(
        equity_hhi[
            [
                "scheme_name",
                "fund_house",
                "sector_hhi_10000"
            ]
        ].head(10)
    )

    print(
        "\nTop 10 most diversified equity funds:"
    )

    display(
        equity_hhi[
            [
                "scheme_name",
                "fund_house",
                "sector_hhi_10000"
            ]
        ]
        .sort_values(
            "sector_hhi_10000"
        )
        .head(10)
    )


# ============================================================
# 27. GENERATE FIVE ADVANCED BUSINESS INSIGHTS
# ============================================================

insights = []


# ------------------------------------------------------------
# INSIGHT 1 - HIGHEST VaR
# ------------------------------------------------------------

if not var_cvar.empty:

    worst_var = var_cvar.iloc[0]

    insights.append(
        f"### 1. Highest Historical VaR\n"
        f"**{worst_var['scheme_name']}** has the most negative "
        f"95% historical daily VaR at "
        f"**{worst_var['VaR_95_pct']:.2f}%**, indicating "
        f"higher downside exposure during the worst 5% "
        f"of observed return days."
    )


# ------------------------------------------------------------
# INSIGHT 2 - HIGHEST CVaR
# ------------------------------------------------------------

if not var_cvar.empty:

    worst_cvar = (
        var_cvar
        .sort_values(
            "CVaR_95_pct"
        )
        .iloc[0]
    )

    insights.append(
        f"### 2. Worst Tail Loss\n"
        f"**{worst_cvar['scheme_name']}** has the most negative "
        f"95% CVaR at "
        f"**{worst_cvar['CVaR_95_pct']:.2f}%**, meaning that "
        f"when returns fall beyond the VaR threshold, the "
        f"average loss is particularly severe."
    )


# ------------------------------------------------------------
# INSIGHT 3 - SIP CONTINUITY
# ------------------------------------------------------------

if len(continuity) > 0:

    at_risk_pct = (
        continuity["at_risk"].mean()
        * 100
    )

    insights.append(
        f"### 3. SIP Continuity Risk\n"
        f"Among investors with at least six SIP transactions, "
        f"**{at_risk_pct:.2f}%** are classified as at-risk "
        f"because they have at least one SIP gap greater than "
        f"35 days."
    )


# ------------------------------------------------------------
# INSIGHT 4 - COHORT INVESTMENT
# ------------------------------------------------------------

if not cohort_summary.empty:

    cohort_for_comparison = (
        cohort_summary
        .dropna(
            subset=[
                "total_invested_all_transactions_inr"
            ]
        )
    )

    if not cohort_for_comparison.empty:

        largest_cohort = (
            cohort_for_comparison
            .sort_values(
                "total_invested_all_transactions_inr",
                ascending=False
            )
            .iloc[0]
        )

        insights.append(
            f"### 4. Strongest Investor Cohort\n"
            f"The **{int(largest_cohort['cohort_year'])}** "
            f"investor cohort has the highest total transaction "
            f"investment at approximately "
            f"**₹{largest_cohort['total_invested_all_transactions_inr']:,.0f}**."
        )


# ------------------------------------------------------------
# INSIGHT 5 - SECTOR CONCENTRATION
# ------------------------------------------------------------

if not equity_hhi.empty:

    highest_hhi = equity_hhi.iloc[0]

    lowest_hhi = (
        equity_hhi
        .sort_values(
            "sector_hhi_10000"
        )
        .iloc[0]
    )

    insights.append(
        f"### 5. Sector Concentration\n"
        f"**{highest_hhi['scheme_name']}** has the highest "
        f"sector HHI among equity funds at "
        f"**{highest_hhi['sector_hhi_10000']:.0f}**, while "
        f"**{lowest_hhi['scheme_name']}** has the lowest at "
        f"**{lowest_hhi['sector_hhi_10000']:.0f}**. "
        f"Higher HHI indicates greater concentration."
    )


# ============================================================
# 28. DISPLAY INSIGHTS AS MARKDOWN
# ============================================================

print(
    "\nADVANCED BUSINESS INSIGHTS"
)

print("=" * 70)

for insight in insights:

    display(
        Markdown(insight)
    )


# ============================================================
# 29. SAVE INSIGHTS AS MARKDOWN FILE
# ============================================================

insights_markdown = (
    "# Advanced Analytics - Business Insights\n\n"
)

for insight in insights:

    insights_markdown += (
        insight + "\n\n"
    )

insights_file = (
    REPORTS / "advanced_analytics_insights.md"
)

with open(
    insights_file,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        insights_markdown
    )

print(
    "\nAdvanced insights saved:"
)

print(insights_file)


# ============================================================
# 30. FINAL DELIVERABLE CHECK
# ============================================================

print("\n")
print("=" * 70)
print("ADVANCED ANALYTICS DELIVERABLE CHECK")
print("=" * 70)

deliverables = [
    var_report_file,
    rolling_sharpe_chart,
    cohort_report_file,
    continuity_report_file,
    recommendation_file,
    recommender_file,
    hhi_report_file,
    insights_file
]

for file in deliverables:

    status = "✓ EXISTS" if file.exists() else "✗ MISSING"

    print(
        f"{status}: {file}"
    )

print("\nAdvanced Analytics completed.")
```
